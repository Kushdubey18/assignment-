Q2.Differentiate between: 
● Series and DataFrame  
Series
A one-dimensional labeled array in Pandas.
Can store data of any type (integer, float, string, etc.).
Similar to a single column in a table.
DataFrame
A two-dimensional tabular data structure with rows and columns.
Similar to an Excel spreadsheet or database table.
Can contain different data types in different columns.

● loc[] and iloc[] functions in Pandas 
loc[] (Label-based Indexing)
Used to access rows and columns by labels (names).
iloc[] (Integer-based Indexing)
Used to access rows and columns by integer positions.


```python
# Q3.
import pandas as pd

# Create DataFrame
data = {
    "Name": ["Amit", "Neha", "Rahul", "Sneha"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 88]
}

df = pd.DataFrame(data)

# Display the DataFrame
print("Original DataFrame:")
print(df)

# Display first 2 rows
print("\nFirst 2 Rows:")
print(df.head(2))

# Display only Name and Marks columns
print("\nName and Marks Columns:")
print(df[["Name", "Marks"]])

# Add a new column named "Result"
df["Result"] = ["Pass", "Pass", "Pass", "Pass"]

print("\nDataFrame with Result Column:")
print(df)

# Display students with marks greater than 80
print("\nStudents with Marks > 80:")
print(df[df["Marks"] > 80])
```

```python
# Q4. 
import pandas as pd
import numpy as np

# Create DataFrame
data = {
    "Name": ["A", "B", "C", "D"],
    "Age": [25, np.nan, 27, 24],
    "Salary": [30000, 40000, np.nan, 35000]
}

df = pd.DataFrame(data)

# Identify missing values
print("Missing Values:")
print(df.isnull())

# Fill missing Age with average age
df["Age"].fillna(df["Age"].mean(), inplace=True)

# Fill missing Salary with median salary
df["Salary"].fillna(df["Salary"].median(), inplace=True)

# Display cleaned dataset
print("\nCleaned Dataset:")
print(df)
```
