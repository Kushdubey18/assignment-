```python
# q1.Write a program that takes a list of numbers and prints all prime numbers in the list using loops. Do not use any built-in prime-checking functions.

# Input list of numbers
numbers = [2, 5, 8, 11, 15, 17, 20, 23, 30]

print("Prime numbers in the list are:")

for num in numbers:
    if num > 1:  # Prime numbers are greater than 1
        is_prime = True

        for i in range(2, num):
            if num % i == 0:
                is_prime = False
                break

        if is_prime:
            print(num)
```

```python
# q3.Write a user-defined function that accepts a list of integers and returns:  
# ● Count of even numbers  
# ● Sum of odd numbers  
# ● Use loops and conditional statements inside the function.


def analyze_numbers(numbers):
    even_count = 0
    odd_sum = 0

    for num in numbers:
        if num % 2 == 0:
            even_count += 1
        else:
            odd_sum += num

    return even_count, odd_sum


# Example list
nums = [10, 15, 20, 25, 30, 35]

even_count, odd_sum = analyze_numbers(nums)

print("Count of even numbers:", even_count)
print("Sum of odd numbers:", odd_sum)  
```

```python
# q4. Create a function that calculates simple interest.  Use default arguments for rate and time. Demonstrate calling the function using both positional and keyword arguments. 



def simple_interest(principal, rate=5, time=1):
    si = (principal * rate * time) / 100
    return si

# Calling with positional arguments
print("Positional Arguments:")
print("SI =", simple_interest(1000))          
print("SI =", simple_interest(1000, 8, 2))   

# Calling with keyword arguments
print("\nKeyword Arguments:")
print("SI =", simple_interest(principal=1000))
print("SI =", simple_interest(principal=1000, rate=10, time=3))
print("SI =", simple_interest(time=2, principal=1000, rate=7))
```

```python
# q5. Create a Student class with attributes: name, marks. Add a method to calculate grade based on marks.


class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def calculate_grade(self):
        if self.marks >= 90:
            return "A+"
        elif self.marks >= 80:
            return "A"
        elif self.marks >= 70:
            return "B"
        elif self.marks >= 60:
            return "C"
        elif self.marks >= 50:
            return "D"
        else:
            return "F"


student1 = Student("Kush", 85)

print("Name:", student1.name)
print("Marks:", student1.marks)
print("Grade:", student1.calculate_grade()) 
```

```python
# q7.  Create a BankAccount class with private attributes for balance. Provide methods to deposit and withdraw money. Ensure balance cannot be accessed directly.



class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   

    def deposit(self, amount):
        self.__balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient Balance!")

    def show_balance(self):
        print("Current Balance:", self.__balance)


account = BankAccount(5000)

account.deposit(1000)

account.withdraw(2000)


account.show_balance()
```

```python
# q8. Write a program that takes two numbers from the user and performs division.  
# ● Handle the following exceptions:  
# ● Division by zero  
# ● Invalid input type 


try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    result = num1 / num2
    print("Result =", result)

except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")

except ValueError:
    print("Error: Invalid input. Please enter numeric values only.") 
```

```python
# q10.  Write a program that reads numbers from a file and calculates:  
# ● Total  
# ● Average  
# ● Handle the case if the file does not exist.


try:
    file = open("numbers.txt", "r")

    numbers = []

    for line in file:
        numbers.append(float(line.strip()))

    file.close()

    total = sum(numbers)
    average = total / len(numbers)

    print("Total =", total)
    print("Average =", average)

except FileNotFoundError:
    print("Error: File does not exist.")
```
